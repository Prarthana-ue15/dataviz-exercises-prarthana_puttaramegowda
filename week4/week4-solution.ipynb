```python
import pandas as pd
import plotly.express as px

# Load Gapminder dataset
df = px.data.gapminder()

print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
print(df.info())

print("Years:", sorted(df["year"].unique()))
print("Continents:", df["continent"].unique())

print(df.describe().round(1))
```

```python
# Filter for Asia and the two selected years
asia = df.loc[
    (df["continent"] == "Asia") &
    (df["year"].isin([2002, 2007]))
].copy()

# Create scatter plot
fig = px.scatter(
    asia,
    x="gdpPercap",
    y="lifeExp",
    color="year",
    color_discrete_map={
        2002: "#9CA3AF",
        2007: "#0072B2"
    },
    hover_name="country",
    hover_data={
        "gdpPercap": ":,.0f",
        "lifeExp": ":.1f",
        "year": True
    },
    log_x=True,
    title="Asian Countries: Life Expectancy Rose as Incomes Grew, 2002–2007",
    labels={
        "gdpPercap": "GDP per Capita (US$; log scale)",
        "lifeExp": "Life Expectancy (years)",
        "year": "Year"
    }
)

# Clean design
fig.update_traces(
    marker=dict(size=10, opacity=0.8)
)

fig.update_layout(
    template="plotly_white",
    title_x=0.02,
    xaxis=dict(showgrid=False, zeroline=False),
    yaxis=dict(showgrid=True, gridcolor="#E5E7EB", zeroline=False),
    legend_title_text="Year"
)

fig.show()
```

```python
# Filter the full dataset for 2007
df_2007 = df.loc[df["year"] == 2007].copy()

# Highlight Europe and grey out all other continents
df_2007["group"] = df_2007["continent"].apply(
    lambda continent: "Europe" if continent == "Europe"
    else "Other continents"
)

# Create bubble chart
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="group",
    color_discrete_map={
        "Europe": "#0072B2",
        "Other continents": "#C7C7C7"
    },
    size_max=45,
    log_x=True,
    hover_name="country",
    hover_data={
        "continent": True,
        "gdpPercap": ":,.0f",
        "lifeExp": ":.1f",
        "pop": ":,.0f",
        "group": False
    },
    title="Europe Stands Out for Higher Incomes and Longer Lives in 2007",
    labels={
        "gdpPercap": "GDP per Capita (US$; log scale)",
        "lifeExp": "Life Expectancy (years)",
        "pop": "Population",
        "group": "Region"
    }
)

# Add annotation explaining the highlighted group
fig.add_annotation(
    x=25000,
    y=79,
    text="Europe: generally higher incomes<br>and longer life expectancy",
    showarrow=True,
    arrowhead=2,
    ax=65,
    ay=-45,
    bgcolor="white",
    bordercolor="#0072B2",
    borderwidth=1,
    font=dict(size=11, color="#0072B2")
)

# Clean design
fig.update_layout(
    template="plotly_white",
    title_x=0.02,
    xaxis=dict(showgrid=False, zeroline=False),
    yaxis=dict(showgrid=True, gridcolor="#E5E7EB", zeroline=False),
    legend_title_text="Region"
)

fig.show()
```
